```python
seller_orders = order_items[['order_id', 'seller_id']].drop_duplicates()

seller_order_count = (
    seller_orders
    .groupby('seller_id')
    .size()
    .reset_index(name='order_count')
)

seller_order_count.head()
```

```python
eligible_sellers = seller_order_count[
    seller_order_count['order_count'] >= 50
].copy()

print("Eligible sellers:", len(eligible_sellers))
eligible_sellers.head()
```

```python
delivered_order_sellers = (
    order_items[['order_id', 'seller_id']]
    .drop_duplicates()
    .merge(
        delivered_orders[['order_id', 'days_late']],
        on='order_id',
        how='inner'
    )
)

delivered_order_sellers.head()
```

```python
seller_late_share = (
    delivered_order_sellers
    .groupby('seller_id')
    .agg(
        delivered_orders=('order_id', 'nunique'),
        late_orders=('days_late', lambda x: (x > 0).sum())
    )
    .reset_index()
)

seller_late_share['late_share'] = (
    seller_late_share['late_orders']
    / seller_late_share['delivered_orders']
)

seller_late_share.head()
```

```python
seller_final = eligible_sellers.merge(
    seller_late_share,
    on='seller_id',
    how='left'
)

seller_final.head()
```

```python
seller_final = seller_final.merge(
    sellers[['seller_id', 'seller_city', 'seller_state']],
    on='seller_id',
    how='left'
)

seller_final.head()
```

```python
print("Number of sellers:", seller_final['seller_id'].nunique())
print("Minimum order count:", seller_final['order_count'].min())
print("Missing seller details:", seller_final['seller_city'].isna().sum())
print("Missing late share:", seller_final['late_share'].isna().sum())
print("Duplicate seller IDs:", seller_final['seller_id'].duplicated().sum())
print("Late share below 0:", (seller_final['late_share'] < 0).sum())
print("Late share above 1:", (seller_final['late_share'] > 1).sum())
```

```python
seller_final.to_csv(
    "seller_late_share.csv",
    index=False
)

print("Saved: seller_late_share.csv")
```
